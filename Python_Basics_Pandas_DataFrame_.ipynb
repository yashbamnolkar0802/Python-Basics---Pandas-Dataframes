**Question 1 :**  Write a Python program to check whether a number is even or odd using conditional statements.

```python
def even_or_odd(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

for num in [4, 7, 0, -3, 10]:
    print(f"{num} is {even_or_odd(num)}")
```

**Question 2 :** Write a Python program to count how many times a number appears in a list.
Example List
[2,3,4,2,5,2]

```python
lst = [2, 3, 4, 2, 5, 2]
target = 2

# Manual loop
count = 0
for item in lst:
    if item == target:
        count += 1
print(f"Loop method    : {target} appears {count} times")

# Built-in
print(f"Built-in method: {target} appears {lst.count(target)} times")
```

**Question 9 :** Create a random array of 10 numbers and:
Find max and min
Sort the array

```python
np.random.seed(42)
rand_arr = np.random.randint(1, 101, size=10)   # 10 integers between 1 and 100

print("Array  :", rand_arr)
print("Max    :", rand_arr.max())
print("Min    :", rand_arr.min())
print("Sorted :", np.sort(rand_arr))
```

**Question 10 :** Replace all even numbers with 0:
arr = np.array([1,2,3,4,5,6])

```python
arr10 = np.array([1, 2, 3, 4, 5, 6])
mask = arr10 % 2 == 0          # [False, True, False, True, False, True]
arr10[mask] = 0
print(arr10)
```
